```python
# Cargar bien desenmascarar los faltante
data_frame = pd.read_csv("empleados_technova_sucio.csv", na_values=["-", "?"])
```

```python
data_frame.isna().sum()   # ahora deberían aparecer los faltantes correctamente contabilizados
```

```python
# Duplicado 
data_frame.duplicated().sum()  # cuántos registros duplicados hay?
```

```python
data_frame = data_frame.drop_duplicates().reset_index(drop=True)  # eliminar duplicados y resetear índices
```

```python
len(data_frame)
```

```python
# Quitar espacio al inicio y al final
for col in ["nombre", "ciudad", "departamento"]:
    data_frame[col] = data_frame[col].str.strip()
```

```python
data_frame.duplicated(subset=["id"]).sum()  # verificar si hay duplicados basados solo en la columna id
```

```python
data_frame[data_frame.duplicated(subset=["id"], keep=False)] # ver las filas que tienen id duplicados para entender mejor el problema
```

```python
data_frame["nombre"] = data_frame["nombre"].str.strip()
data_frame["ciudad"] = data_frame["ciudad"].str.strip()
data_frame["departamento"] = data_frame["departamento"].str.strip()
```

```python
data_frame = data_frame.str.strip()
```

```python
# Detectar automaticamente las columnas de texto
cols_texto = data_frame.select_dtypes(include=["str"]).columns
# aplicar strip a solo esas
data_frame[cols_texto] = data_frame[cols_texto].apply(lambda x: x.str.strip())
```

```python
# Cargar bien desenmascarar los faltante
data_frame = pd.read_csv("empleados_technova_sucio.csv", na_values=["-", "?"])
```

```python
data_frame.isna().sum()   # ahora deberían aparecer los faltantes correctamente contabilizados
```

```python
# Duplicado 
data_frame.duplicated().sum()  # cuántos registros duplicados hay?
```

```python
data_frame = data_frame.drop_duplicates().reset_index(drop=True)  # eliminar duplicados y resetear índices
```

```python
len(data_frame)
```

```python
# Quitar espacio al inicio y al final
for col in ["nombre", "ciudad", "departamento"]:
    data_frame[col] = data_frame[col].str.strip()
```

```python
data_frame.duplicated(subset=["id"]).sum()  # verificar si hay duplicados basados solo en la columna id
```

```python
data_frame[data_frame.duplicated(subset=["id"], keep=False)] # ver las filas que tienen id duplicados para entender mejor el problema
```

```python
data_frame["nombre"] = data_frame["nombre"].str.strip()
data_frame["ciudad"] = data_frame["ciudad"].str.strip()
data_frame["departamento"] = data_frame["departamento"].str.strip()
```

```python
#data_frame = data_frame.str.strip()
```

```python
# Detectar automaticamente las columnas de texto
cols_texto = data_frame.select_dtypes(include=["str"]).columns
# aplicar strip a solo esas
data_frame[cols_texto] = data_frame[cols_texto].apply(lambda x: x.str.strip())
```

```python
display(data_frame)
```

```python
data_frame["ciudad"] = data_frame["ciudad"].str.title()  # "santiago" -> "Santiago"
data_frame["departamento"] = data_frame["departamento"].str.capitalize()  # "ventas" -> "Ventas"
```

```python
data_frame["ciudad"].unique() # cuantos valores únicos hay en la columna ciudad?
data_frame["departamento"].unique() # cuantos valores únicos hay en la columna departamento?
```

```python
data_frame["salario"] = (data_frame["salario"].str.replace("$","", regex=False)
                                              .str.replace(".","", regex=False)
                                              .astype(float))
```

```python
data_frame["salario"].dtype # float64, ahora sí es numérica y se pueden hacer cálculos con ella
```

```python
display(data_frame)
```

```python
print("edad antes:  ", sorted(data_frame["edad"].dropna().unique().astype(int)))
```

```python
# regla de negocio: edad válida entre 18 y 90
data_frame.loc[(data_frame["edad"] < 18) | (data_frame["edad"] > 90), "edad"] = np.nan
```

```python
print("edad después:", sorted(data_frame["edad"].dropna().unique().astype(int)))
```

```python
display(data_frame)
```

```python
# Tratar los valores perdidos
# edad -> mediana (numérica y robusta ante atípicos)
data_frame["edad"] = data_frame["edad"].fillna(data_frame["edad"].median()).astype(int)
```

```python
# salario -> mediana
data_frame["salario"] = data_frame["salario"].fillna(data_frame["salario"].median())
```

```python
data_frame["correo"] = data_frame["correo"].fillna("sin_correo")
```
